```python
import pandas as pd
```

```python
df = pd.read_excel("D:\Voltride_Analysis_DecodeX\modified_voltride_dataset_DecodeX.xlsx")
df.head()
```

```python
df["Revenue"] = df.apply(
    lambda x: x["Estimated_Fare"] if x["Ride_Status"] == "Completed" else 0,
    axis=1
)

df.head()
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
driver_payout = total_revenue * 0.70

print("Driver Payout (70%):", round(driver_payout, 2))
```

```python
completed_rides = df[df["Ride_Status"] == "Completed"].shape[0]

charging_cost = completed_rides * 20

print("Charging Cost:", charging_cost)
```

```python
net_profit = total_revenue - driver_payout - charging_cost

print("Net Profit:", round(net_profit, 2))
```

```python
profit_margin = (net_profit / total_revenue) * 100

print("Profit Margin (%):", round(profit_margin, 2))
```

```python
df["Month_Name"] = df["Date"].dt.month_name()

monthly_revenue = df.groupby("Month_Name")["Revenue"].sum().reset_index()

monthly_revenue
```

```python
monthly_revenue["Driver_Payout"] = monthly_revenue["Revenue"] * 0.70

monthly_revenue["Charging_Cost"] = (
    df[df["Ride_Status"] == "Completed"]
    .groupby(df["Date"].dt.month_name())
    .size()
    .values * 20
)

monthly_revenue["Net_Profit"] = (
    monthly_revenue["Revenue"]
    - monthly_revenue["Driver_Payout"]
    - monthly_revenue["Charging_Cost"]
)

monthly_revenue
```

# save the dataframe as Excel file

```python
monthly_revenue.to_excel("Financial_Performance_Report.xlsx", index=False)
```
